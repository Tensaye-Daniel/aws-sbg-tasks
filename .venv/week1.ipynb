```python
import numpy as np 
import pandas as pd

df= pd.read_csv("data/students.csv")
print(df)
```

```python
print(df.info())
```

```python
#cleaning Function
def remove_duplicates(data):
    return data.drop_duplicates()
def clean_missing_values(data):
     data= data.copy()
     data["age"]=data["age"].fillna(data["age"].median())
     data["score"]= data["score"].fillna(data["score"].mean())
     return data

def clean_data(data):
    data=remove_duplicates(data)
    data=clean_missing_values(data)
    return data
```

```python
cleanned_df=clean_data(df)
print(cleanned_df)
print(cleanned_df.isnull().sum())
```
