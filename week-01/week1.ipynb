```python
import numpy as np 
import pandas as pd

df= pd.read_csv("week-01/data/students.csv")
print(df)
```

```python
print(df.info())
```

```python
#cleaning Function
def remove_duplicates(data):
    return data.drop_duplicates()
def clean_missing_values(data):
     data= data.copy()
     data["age"]=data["age"].fillna(data["age"].median())
     data["score"]= data["score"].fillna(data["score"].mean())
     return data

def clean_data(data):
    data=remove_duplicates(data)
    data=clean_missing_values(data)
    return data
```

```python
cleanned_df=clean_data(df)
print(cleanned_df)
print(cleanned_df.isnull().sum())
```

```python
scores = cleanned_df["score"].to_numpy()

average_score = np.mean(scores)
highest_score = np.max(scores)
lowest_score = np.min(scores)
standard_deviation = np.std(scores)

print("Average Score:", round(average_score, 2))
print("Highest Score:", highest_score)
print("Lowest Score:", lowest_score)
print("Standard Deviation:", round(standard_deviation, 2))
```

```python
def classify_students(data):
    data = data.copy()

    data["result"] = np.where(
        data["score"] >= 50,
        "Pass",
        "Fail"
    )

    return data


final_df = classify_students(cleanned_df)

print("Final Processed Dataset:")
print(final_df)
```

```python
final_df.to_csv("data/cleaned_students.csv", index=False)

print("Cleaned dataset saved successfully!")
```
